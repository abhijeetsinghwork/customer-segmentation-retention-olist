# Data Understanding & Exploration

This notebook explores the structure, relationships, and quality of the Olist e-commerce dataset.
The goal is to understand the data before feature engineering and modeling.

This cell imports the core Python libraries required for data analysis.  
Pandas is used for data manipulation and working with tabular datasets, NumPy provides numerical computing utilities, and Path from the pathlib module helps manage file paths in a platform-independent way.  

The display setting is also modified so that pandas shows all columns of a dataframe when it is printed, which makes exploratory analysis easier.

```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
```

This cell loads all the raw CSV files from the dataset into pandas DataFrames.  
A central data directory is defined so that file paths remain clean and easy to manage throughout the project.  

Each CSV file represents a different table in the Olist e-commerce dataset such as customers, orders, products, sellers, payments, and reviews. After loading the data, the number of rows in each table is printed to quickly confirm that all files were loaded successfully.

```python
DATA_DIR = Path("../data/raw")

customers = pd.read_csv(DATA_DIR / "olist_customers_dataset.csv")
geolocation = pd.read_csv(DATA_DIR / "olist_geolocation_dataset.csv")
order_items = pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv")
payments = pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv")
reviews = pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv")
orders = pd.read_csv(DATA_DIR / "olist_orders_dataset.csv")
products = pd.read_csv(DATA_DIR / "olist_products_dataset.csv")
sellers = pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv")
category_translation = pd.read_csv(DATA_DIR / "product_category_name_translation.csv")

print("Loaded tables:",
      len(customers), len(geolocation), len(order_items), len(payments),
      len(reviews), len(orders), len(products), len(sellers), len(category_translation))
```

This cell defines a helper function to quickly inspect the structure and quality of each dataset.  

For every dataframe, the function prints the table name, shows the first few rows, displays the dataset shape, and highlights the columns with the most missing values.  

This allows us to quickly understand the schema of each table and identify potential data quality issues before moving on to feature engineering and modeling.

```python
def inspect_df(df, name, n=5):
    print(f"\n--- {name} ---")
    display(df.head(n))
    print(df.shape)
    display(df.isna().sum().sort_values(ascending=False).head(15))

for name, df in {
    "customers": customers,
    "geolocation": geolocation,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}.items():
    inspect_df(df, name)
```
